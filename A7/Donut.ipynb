```python
# Identify the three provinces or territories with the largest raw Indigenous_Total values (Ontario, Quebec, Alberta). For those three
# locations, create one row of three Plotly donut subplots showing composition within the Indigenous
# population only: First Nations, Inuit, and Métis.


# - Order locations from largest to smallest Indigenous_Total.
# - Use the same group labels and colors used elsewhere; use spaces rather than underscores.
# - Show percentages on the donuts, placing them inside where possible and outside when necessary.
# - Place a centered annotation above each donut with the location name and Indigenous total count.
# - Use one shared legend. Format each tooltip to show only the group name, raw count, and
# percentage within the Indigenous population.
# - Save donuts.html.

# - Add this to the webpage created in Part 3.
# - In a few sentences, compare the Indigenous composition of the three locations.

data['First Nations'] = data['First_Nations']
data['Indigenous Total'] = data['First Nations'] + data['Inuit'] + data['Métis']

ontario_data = data[data['Location'] == 'Ontario']
quebec_data = data[data['Location'] == 'Quebec']
alberta_data = data[data['Location'] == 'Alberta']


labels = ['First Nations', 'Inuit', 'Métis']
```

```python
from plotly.subplots import make_subplots
fig = make_subplots(rows=1, cols=3, specs=[[{"type": "domain"}, {"type": "domain"}, {"type": "domain"}]])



fig.add_trace(
    go.Pie(
        labels=labels,
        values=ontario_data[["First Nations", "Inuit", "Métis"]].sum(),
        hole=0.5,
        marker=dict(
            colors=["green", "purple", "blue"],
            line=dict(color="#000000", width=1),
        ),
        textposition="inside",
        textinfo="label+percent",
        textfont_size=16,
        hoverinfo="value",
        hovertemplate=(
            '<span style="background-color:grey;color:black;'
            'font-family:Georgia;font-weight:bold;font-size:16px;">'
            '%{label}<br><br>%{value:,}<br>%{percent:.1%}</span><extra></extra>'
        ),
        hoverlabel=dict(bgcolor="lightgrey"),
    ),
    col=1,
    row = 1
)


fig.add_trace(
    go.Pie(
        labels=labels,
        values=quebec_data[["First Nations", "Inuit", "Métis"]].sum(),
        hole=0.5,
        marker=dict(
            colors=["green", "purple", "blue"],
            line=dict(color="#000000", width=1),
        ),
        textposition="inside",
        textinfo="label+percent",
        textfont_size=16,
        hoverinfo="value",
        hovertemplate=(
            '<span style="background-color:grey;color:black;'
            'font-family:Georgia;font-weight:bold;font-size:16px;">'
            '%{label}<br><br>%{value:,}<br>%{percent:.1%}</span><extra></extra>'
        ),
        hoverlabel=dict(bgcolor="lightgrey"),
    ),
    col=2,
    row = 1
)


fig.add_trace(
    go.Pie(
        labels=labels,
        values=alberta_data[["First Nations", "Inuit", "Métis"]].sum(),
        hole=0.5,
        marker=dict(
            colors=["green", "purple", "blue"],
            line=dict(color="#000000", width=1),
        ),
        textposition="inside",
        textinfo="label+percent",
        textfont_size=16,
        hoverinfo="value",
        hovertemplate=(
            '<span style="background-color:grey;color:black;'
            'font-family:Georgia;font-weight:bold;font-size:16px;">'
            '%{label}<br><br>%{value:,}<br>%{percent:.1%}</span><extra></extra>'
        ),
        hoverlabel=dict(bgcolor="lightgrey"),
    ),
    col = 3,
    row = 1
)

totals = [
    ontario_data["Indigenous Total"].iloc[0],
    quebec_data["Indigenous Total"].iloc[0],
    alberta_data["Indigenous Total"].iloc[0]
]
locations = [
    ontario_data["Location"].iloc[0],
    quebec_data["Location"].iloc[0],
    alberta_data["Location"].iloc[0]
]


fig.update_layout(
    annotations=[
        dict(
            text=f"<b>{location}<br>Indigenous Total:<br>{total}</b>",
            x=x,
            y=1.01,
            font=dict(size=14, color="grey"),
            showarrow=False,
        )
        for x, total, location in zip([1 / 13, 0.5, 12 / 13], totals, locations)
    ],
    height=600,
    width=1200,
    showlegend=True,
    legend=dict(
        orientation="h",
        yanchor="top",
        y=-0.05,
        xanchor="center",
        x=0.5,
    ),
)

fig.write_html("Donut.html")
fig.show()
```
