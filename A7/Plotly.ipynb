# Assignment 7 Part 2 - Scatter Plots

```python
import plotly.express as px
import pandas as pd

path = 'Data/Canada Merge Data.csv'
```

```python
data = pd.read_csv(path)
data.info()
```

```python
data = data[data['Domain'] != 'Missing']
data['Domain'] = data['Domain'].replace('NonIndigenous [reference]', 'Non Indigenous')
data['Domain'] = data['Domain'].replace('First Nations off reserve', 'First Nations')
data
```

```python
max_x,max_y = data['CR_BELONG'].max(), data['CR_HSC'].max()
max_x *= 1.05
max_y *= 1.05
min_x, min_y = 0,0

colors = {
    'Non Indigenous': 'lightgray',
    'First Nations': 'forestgreen',
    'Inuit': 'mediumpurple',
    'Métis': 'royalblue'
}

symbols = {
    'Non Indigenous': 'circle',
    'First Nations': 'diamond',
    'Inuit': 'square',
    'Métis': 'triangle-up'
}

fig = px.scatter(
    data, x = 'CR_BELONG',
    y = 'CR_HSC', hover_name = 'Geography',
    color = 'Domain', symbol= 'Domain',
    animation_frame = 'Sex', 
    color_discrete_map = colors, symbol_map = symbols,
    labels = {
        'CR_BELONG':  'Percentage Reporting Strong or Somewhat Strong Community Belonging ',
        'CR_HSC': 'Percentage Reporting all Perceived Mental-Health Needs were Met '
    },
    range_x = [min_x, max_x], range_y = [min_y, max_y],
    hover_data = {
        'Domain': False, 'Sex': False, 'CR_BELONG': False, 'CR_HSC': False
        },
    custom_data= ['DENOM_BELONG', 'DENOM_HSC'],
    width = 800, height = 800
)

fig.update_traces(marker=dict(size=10))
fig.update_traces(
    hovertemplate=
    "<b>%{hovertext}</b><br>" +
    "Those who feel Community Belonging: %{x:.2f}%<br>" +
    "Those Saying Mental Health Needs are Met: %{y:.2f}%<br>" +
    'Belonging Survey Respondents: %{customdata[0]:,.0f}<br>' +
    'Mental Health Survey Respondents: %{customdata[1]:,.0f}' +
    '<extra></extra>'
)

fig.write_html('scatter.html')
fig.show()
```
